```python
import pandas as pd
import numpy as np
```

```python
DATA_PATH = "../data/raw/telco_customer_churn.csv"
```

```python
df = pd.read_csv(DATA_PATH)
print(f"Original shape: {df.shape}")
```

```python
TARGET = "Churn Label"
y = df[TARGET].copy()
X = df.drop(columns=[TARGET]).copy()
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
columns_to_drop = [
    "Count",
    "Country",
    "State",
    "CustomerID",
    "Churn Value",
    "Churn Score",
    "Churn Reason"
]
X = X.drop(columns=columns_to_drop)
```

```python
print("Remaining features:")
print(X.columns.tolist())
print(f"\nFeature count: {X.shape[1]}")
```

```python
X["Total Charges"] = pd.to_numeric(
    X["Total Charges"],
    errors="coerce"
)
```

```python
print("Missing Total Charges:", X["Total Charges"].isna().sum())
```

```python
X[X["Total Charges"].isna()]
```

```python
X.dtypes
```

```python
print("\nNumerical features:")
print(X.select_dtypes(include=np.number).columns.tolist())
print("\nCategorical features:")
print(X.select_dtypes(include=["object", "string"]).columns.tolist())
```

```python
missing = (
    X.isnull().sum().sort_values(ascending=False)
)
missing[missing > 0]
```

```python
print(y.value_counts())
print("\nTarget percentage:")
print(y.value_counts(normalize=True) * 100)
```

```python
X.loc[(X["Tenure Months"] == 0) & (X["Total Charges"].isna()),"Total Charges"] = 0
```

```python
print("Missing Total Charges after cleaning:",X["Total Charges"].isna().sum())
print("\nZero-tenure customers:")
print((X["Tenure Months"] == 0).sum())
print("\nRemaining missing values:")
print(X.isnull().sum()[X.isnull().sum() > 0])
```
