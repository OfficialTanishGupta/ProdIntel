```python
import pandas as pd
import numpy as np
```

```python
DATA_PATH = "../data/raw/telco_customer_churn.csv"
df = pd.read_csv(DATA_PATH)
print("Original shape:", df.shape)
```

```python
TARGET = "Churn Label"
y = df[TARGET].copy()
X = df.drop(columns=[TARGET]).copy()
```

```python
columns_to_drop = ["Count","Country","State","CustomerID","Churn Value","Churn Score","Churn Reason"]
```

```python
X = X.drop(columns=columns_to_drop)
X["Total Charges"] = pd.to_numeric(X["Total Charges"],errors="coerce")
X.loc[(X["Tenure Months"] == 0) & (X["Total Charges"].isna()),"Total Charges"] = 0
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
y = y.map({"No": 0,"Yes": 1})
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
print("Training target distribution:")
print(y_train.value_counts(normalize=True) * 100)
print("\nTest target distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
print("Training rows:", len(X_train))
print("Testing rows :", len(X_test))
print("\nTraining indexes:", X_train.index[:5].tolist())
print("Testing indexes :", X_test.index[:5].tolist())
print("\nOverlapping rows:",len(set(X_train.index).intersection(set(X_test.index))))
```

```python
print("Missing values in X_train:")
print(X_train.isnull().sum()[X_train.isnull().sum() > 0])
print("\nMissing values in X_test:")
print(X_test.isnull().sum()[X_test.isnull().sum() > 0])
```
